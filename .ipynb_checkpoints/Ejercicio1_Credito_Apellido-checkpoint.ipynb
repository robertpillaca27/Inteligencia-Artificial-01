# Ejercicio 1: Evaluación de Solicitud de Crédito Bancario

### Ficha PEAS
* **Medida de desempeño (P):** Tasa de retorno de capital, porcentaje de créditos impagos y tiempo de evaluación por solicitud.
* **Entorno (E):** Sistema financiero, base de datos del banco y buró de crédito.
* **Acciones (A):** `Aprobar`, `Aprobar con condiciones`, `Rechazar`.
* **Percepción (S):** `ingreso_mensual`, `monto_solicitado`, `tiene_historial_moroso`.

### Justificación de las reglas
Un historial moroso representa un alto riesgo de incumplimiento; sin embargo, si la deuda solicitada es baja en comparación con sus ingresos ($\le 0.4$), se puede otorgar con condiciones y garantías adicionales. Para clientes sin historial negativo, una relación monto/ingreso menor al $30\%$ demuestra excelente capacidad de pago; entre $30\%$ y $60\%$ compromete su flujo de caja requiriendo mayor tasa; y más del $60\%$ representa un sobreendeudamiento irresponsable para el banco.
